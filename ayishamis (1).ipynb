```python
plt.figure(figsize=(10, 5))

d1["destination_country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Destination Countries")
plt.xlabel("Destination country")
plt.ylabel("Number of Students")
plt.legend()
```

```python
plt.figure(figsize=(10, 5))

d1["origin_country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 origin Countries")
plt.xlabel("origin country")
plt.ylabel("Number of Students")
plt.legend()
```

```python
plt.figure(figsize=(10, 5))

d1["destination_country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Destination Countries")
plt.xlabel("Destination country")
plt.ylabel("Number of Students")
plt.legend()

plt.title('Reasons for choising UAE')
plt.xlabel('Reasons')
plt.ylabel('Number of students')
plt.show()
```

```python
uae=d1[d1['destination_country'] =='UAE']
uae['enrollment_reason'].value_counts().plot(kind="bar")

plt.title('Reasons for Choosing UAE')
plt.xlabel('Reason')
plt.ylabel('Number of Students')
plt.show()
```

```python
d1['placement_status'].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.title("placed vs not placed")
plt.ylabel("  ")
plt.legend()
plt.show()
```

```python
d1.groupby('placement_country')['starting_salary_usd'].mean().sort_values().plot(kind='bar')
plt.title('Placement Country vs Average Starting Salary')
plt.xlabel('Placement Country')
plt.ylabel('Average Starting Salary (USD)')
plt.show()
```

```python
sns.barplot(data=d1, x='field_of_study', y='starting_salary_usd')
plt.title('Field of Study vs Starting Salary')
plt.xticks(rotation=45)
plt.show()
```

```python
pd.crosstab(d1['origin_country'], d1['destination_country']).plot(kind='bar', stacked=True, figsize=(12,6))
plt.title('Origin Country vs Destination Country')
plt.xlabel('Origin Country')
plt.ylabel('Number of Students')
plt.show()
```

```python
sns.countplot(data=d1, x='origin_country',
              hue='scholarship_received')
plt.title('Scholarship vs Origin Country')
plt.xticks(rotation=45)
plt.show()
```

```python
d1.groupby('placement_status')['gpa_or_score'].mean().sort_values().plot(kind='bar')
plt.title('Placement_status vs Average Starting Sala')
plt.xlabel('Placement status')
plt.ylabel('gpa_or_score')
plt.show()
```
